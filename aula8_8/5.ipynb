# Contagem Simples, o usúario digita um número e o programa imprime do zero até o número
#1

```python
n = int(input("Digite um Número: "))

for i in range(-1, n):
  print(i + 1)
```

#Contagem Regressiva, o usuário digita um número e o programa imprime do número até o zero

#2

```python
n = int(input("Digite um Número: "))

for i in range(n, -1, -1):
  print(i)
```

# Contar vogais, usúario digita uma palavra e o programa conta as vogais

#8

```python
palavra = input("Digite uma palavra: ")

vogais = 0

for letra in palavra.lower():
    if letra in "aeiou":
        vogais += 1

print(f"A quantidade de vogais na palavra {palavra}, é igual a: {vogais}")
```

#Contar consoantes, usúario digita uma palavra e o programa conta as consoantes

#8,5


```python
palavra = (input("Digite uma palavra: "))

consoantes = 0

for letra in palavra.lower():
  if letra in "bcdfghjklmnpqrstvwxyz":
    consoantes += 1

print(f"A quantidade de consoantes na palavra {palavra}, é igual a: {consoantes}")
```

# Maior número digitado, o usuáro digita quatro números e o programa imprime o maior número digitado

#O usúario digita um número entre 01 e 100 e o programa mostra quais são divisíveis por 3 e 5
