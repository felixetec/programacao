```python
cor = input("Digite uma cor: ").lower()

match cor:
    case "azul":
        print("Cor azul")
    case _:
        print("Outra cor")
```

```python
numero = int(input("Digite um número (1, 2 ou 3): "))

match numero:
    case 1:
        print("Um")
    case 2:
        print("Dois")
    case 3:
        print("Três")
```

```python
numero = int(input("Escolha uma bebida (1, 2 ou 3): "))

match numero:
    case 1:
        print("Água")
    case 2:
        print("Suco")
    case 3:
        print("Refrigerante")
    case _:
        print("Opção inválida")
```

```python
dia = input("Digite um dia da semana: ").lower()

match dia:
    case "sábado" | "sabado" | "domingo":
        print("Final de semana")
    case _:
        print("Dia de semana")
```

```python
nota = input("Digite a nota (A, B ou C): ").upper()

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")
```
