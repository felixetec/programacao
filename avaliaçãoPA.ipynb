QUESTÃO 02 – Soma dos Números (2 pontos)
---------------------------------------------------------

Solicite ao usuário um número inteiro positivo.

Utilizando um laço de repetição, calcule e exiba a soma de
todos os números de 1 até o número informado.

Exemplo:

Digite um número: 5

Resultado: 15

Pois:

# 1 + 2 + 3 + 4 + 5 = 15

```python
soma = 0
n = int(input("Digite um número: "))
for i in range(n + 1):
  soma += i
  print(n + soma - n)
```

#Solicite ao usuário um número inteiro positivo e calcule
seu fatorial utilizando um laço de repetição.

Exemplo:

Digite um número: 5

Fatorial = 120

Pois:

5! = 5 × 4 × 3 × 2 × 1

```python
n = int(input("digite um número: "))

if n < 0:
    print("fatorial não existe para números negativos.")
elif n == 0:
    print("fatorial de 0 é 1.")
else:
    factorial = 1
    for i in range(1, n + 1):
        factorial *= i
    print(f"fatorial = {factorial}")
```

QUESTÃO 05 – Caixa Eletrônico (5 pontos)
---------------------------------------------------------

Um caixa eletrônico possui notas de:

R$ 100
R$ 50
R$ 20
R$ 10
R$ 5
R$ 2
R$ 1

Solicite ao usuário o valor do saque e informe a menor
quantidade possível de notas para realizar o pagamento.

Exemplo:

Digite o valor do saque: 287

Saída esperada:

2 nota(s) de R$100
1 nota(s) de R$50
1 nota(s) de R$20
1 nota(s) de R$10
1 nota(s) de R$5
1 nota(s) de R$2

Requisitos:
- Utilizar laços de repetição.
- Não utilizar listas.
- Exibir apenas as notas utilizadas.

```python
saque = int(input("digite o valor que deseja sacar: "))
n = (100, 50, 20, 10, 5, 2, 1)
print('')
for nota in n:
  q = saque // nota
  if q > 0:
    print(f"{q} nota(s) de R${nota}")
    saque %= nota
```
