```python
for i in range(5):
  print(i)
```

```python
for i in range(0, 11, 2):
  print(i)
```

```python
for i in range(10, -1, -1):
  print(i)
```

```python
n = 2

match n:
  case 1:
    print('Você escolheu o número 1')
  case 2:
      print('Você escolheu o número 2')
  case 3:
      print('Você escolheu o número 3')
  case _:
    print('Você escolheu um número inválido')
```

```python
opcao = 1

match opcao:
  case 1:
    print('Hambúrguer')
  case 2:
    print('Pizza')
  case 3:
    print('Hot Dog')
  case _: # o underline serve para reconhecer qualquer qualquer outra opção que nao tenha no case
    print('Você escolheu uma opção inválida')
```

```python
dia = 'Sábado'

match dia:
  case 'Segunda':
    print('Dia difícil')
  case 'Terça':
    print('Bom')
  case 'Quarta':
    print('Meio da semana')
  case 'Quinta':
    print('top')
  case 'Sexta':
    print('Ok')
  case 'Sábado':
    print('Perfeito')
  case 'Domingo':
    print('Lindo')
  case _:
    print('Você escolheu um dia inválido')
```
