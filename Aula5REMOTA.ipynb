1- Soma simples

#Faça um programa que:

* Leia dois números inteiros
* Mostre a soma deles

```python
#Entrada

numero_1 = float (input("Digite o primeiro número: "))
numero_2 = float (input("DIgite o segundo número: "))

#Processamento

soma = numero_1 + numero_2

#Saída
print(f"A soma dos dois números é igual a: {soma}")
```

5- Média de 3 números

#Crie um algorítimo que:

* Leia três números
* Calcule e mostre a média


```python
#Entrada

numero_1 = float(input("Digite o primeiro número: "))
numero_2 = float(input("Digite o segundo número: "))
numero_3 = float(input("Digite o terceiro número: "))

#Processamento

media = (numero_1 + numero_2 + numero_3) / 3

#Saída
print(f"A média dos números é igual a: {media}")
```

6 - Divisão inteira e resto

#Faça um programa que:

* Leia um número inteiro
* Mostre:
* O dobro
* A metade
* O resto da divisão por 3


```python
#Entrada

numero = float(input("Digite o número: "))

#Processamento

dobro = numero * 2
metade = numero / 2
resto = numero % 3

#Saída
print(f"O dobro do número é igual a: {dobro}")
print(f"A metade do número é igual a: {metade}")
print(f"O resto do número é igual a: {resto}")

```

7 - Conversão de Temperatura

#Desenvolva um programa que:
* Leia a temperatura em Celsius
* Converta para Fahrenheit


```python
#Entrada

c = float(input("digite a temperatura em graus celsius: "))

#Processamento

fahrenheit = c * (9 / 5) + 32

#Saída

print(f"A temperatura de firenheit é igual a: {fahrenheit}")
```

8 - Cálculo do salário

#Desenvolva um programa que:

* Leia o salário de um funcionário
* Calcule o aumento de 10%
* Mostre o novo salário



```python
#Entrada
salario = float(input("Digite o salário do funcionário: "))

#Processamento
aumento_do_salario = salario + (salario * 0.10)

#Saída
print(f"O novo salário do funcionario aumentou para: {aumento_do_salario}")
```

10 - Consumo de combustível

#Faça um programa que:

* Leia a distância percorrida
* Leia o combustível gasto (litros)
* Calcule o consumo médio (km/l)

```python
#Entrada

distancia = float(input("Digite a distância percorrida do carro (km): "))
combustivel = float(input("Digite o combustível gasto (litros): "))

#Processamento
consumo = distancia / combustivel

#Saída
print(f"O carro consumiu: {consumo} litros")

```
