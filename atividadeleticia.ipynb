Exercício 1 – Idade para dirigir
Receba a idade de uma pessoa e diga se ela pode tirar carteira de motorista. A pessoa
pode tirar a carteira se tiver idade maior ou igual a 18 anos.

```python
idade = int(input("Digite sua idade: "))

if idade >= 18:
    print("Você já pode tirar a carteira de motorista.")
else:
    print("Você ainda não pode tirar a carteira de motorista.")
```

# Receba um número e diga se ele é par ou ímpar.

```python
n = int(input('Digite um número: '))

if n % 2 == 0:
  print(f'O número {n} é par')

else:
  print(f"O número {n} é impar")
```

E\xercício 3 – Maior de dois números
Receba dois números e mostre qual é o maior. Se os dois números forem iguais, diga que
eles são iguais.

```python
n1 = float(input("Digite o primeiro número: "))
n2 = float(input("Digite o segundo número: "))

if n1 > n2:
    print(f"O maior número é: {n1}")
elif n2 > n1:
    print(f"O maior número é: {n2}")
else:
    print("Os dois números são iguais.")
```

Exercício 4 – Situação do aluno
Receba a nota de um aluno e mostre sua situação:
 Nota maior ou igual a 7 → Aprovado
 Nota maior ou igual a 5 e menor que 7 → Recuperação
 Nota menor que 5 → Reprovado


```python
nota = float(input("Digite a sua nota: "))

if nota >= 7.0:
    print("Aprovado")
elif nota >= 5.0 and nota < 7.0:
    print("Recuperação")
else:
    print("Reprovado")
```

Exercício 5 – Desconto na compra
Receba o valor de uma compra. Se o valor for maior ou igual a 100, aplique 10% de
desconto e mostre o valor final. Caso contrário, mostre o valor normal da compra.

Exercício 6 – Positivo, Negativo ou Zero
Receba um número e diga se ele é positivo, negativo ou zero.


```python
n = float(input('Digite um número: '))

if n > 0:
  print(f'O número {n} é positivo')

elif n < 0:
  print(f"O número {n} é negativo")

else:
  print(f"O número {n} é zero")
```

Exercício 7 – Classificação de idade
Receba a idade de uma pessoa e classifique-a de acordo com sua faixa etária:
 Menor de 12 anos → Criança
 De 12 a 17 anos → Adolescente
 De 18 a 59 anos → Adulto
 60 anos ou mais → Idoso

```python
idade = int(input('Digite sua idade: '))

if idade < 12:
  print(f"Você é um criança")
elif idade >=12 and idade <= 17:
  print(f"Você é um adolescente")
elif idade >= 18 and idade <=59:
  print(f"Você é um adulto")
else:
  print(f'Você é um idoso')
```

Exercício 8 – Média de duas notas
Receba duas notas, calcule a média e diga se o aluno foi aprovado ou reprovado. O aluno
será aprovado se a média for maior ou igual a 7.

```python
n1 = float(input('Digite sua primeira nota: '))
n2 = float(input('Digite sua segunda nota: '))

media = (n1 + n2) / 2

if media >= 7:
  print('Você foi aprovado')
else:
  print('Você foi reprovado')
```

Receba o peso e a altura de uma pessoa, calcule o IMC usando a fórmula:
IMC = peso / (altura × altura)
Depois, classifique o resultado:
 IMC menor que 18.5 → Abaixo do peso
 IMC de 18.5 a 24.9 → Peso normal
 IMC de 25 a 29.9 → Sobrepeso
 IMC maior ou igual a 30 → Obesidade

```python
altura = float(input('Digite sua altura: '))
peso = float(input('Digite seu peso: '))

imc = peso / (altura * altura)

if imc < 18.5:
    print(f"Você está abaixo do peso")
elif imc >= 18.5 and imc <= 24.9:
    print(f"Seu peso está normal")
elif imc >= 25 and imc <= 29.9:
    print(f"Você está acima do peso")
else:
    print(f"Você está com gigantesco")
```
