Exercício 3 – Conversões
Faça as conversões e mostre o resultado: Python

texto1 = "20" texto2 = "3.5" numero = 8

a) Transforme texto1 em inteiro e some com 10 b) Transforme texto2 em real e multiplique por 2 c) Transforme número em texto e junte com a palavra "anos"

```python
# CRIANDO A VARIÁVEL
texto1 = "20"
texto2 = "3.5"
numero = 8

#Transformando
result_a = int(texto1) + 10
result_b = float(texto2) * 2
result_c = str(numero)

# Saída

print (f"O resultado da letra a é: {result_a}.")
print (f"O resultado da letra b é: {result_b}.")
print (f"O resultado da letra c é: {result_c} anos.")
```

Exercício 6 – Situação prática
Um produto custa R$ 80,00. O cliente ganhou 10% de desconto. Faça um programa que calcule e mostre: • O valor do desconto • O valor final a pagar • Se o valor final é menor que 75.00 (True ou False)

```python
produto = 80
desconto = (produto * 10)/100
valor = produto - desconto

#Saída

print (f"O valor do desconto é de R${desconto}. ")
print (f"O valor final é de R${valor}.")
print (f"O valor final é menor que 75?", valor < 75)
```
