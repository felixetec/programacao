```python
# Estrutura de Decisão

n = 5

n1 = int (input("Digite um número: "))
print('-=-'*10)

if n > n1:
  print(f"{n} é maior que {n1}")

elif n == n1:
  print(f"{n} é igual ao numero: {n1}")

else:
  print(f"{n} é menor que {n1}")

print('-=-'*10)
```

```python
# Escrever um programa que leia dois números e retorne se eles são iguais
print('-=-'*10)
a = int (input("Digite um número: "))
b = int (input("Digite um número:"))
print('-=-'*10)

if a > b:
  print(f"O número {a} é maior que {b}")

elif a == b:
  print(f"o número {a} é igual ao número: {b}")

else:
  print(f"O número {a} é menor que {b}")

print('-=-'*10)
```

```python
# Escrever um programa que leia as notas dos quatro bimestres, nome do aluno e calcule a média

# Média menor que 5 - Reprovado
# Média menor que 7 - Recuperação
# Média Maior que 7 - Aprovado

print('-=-'*10)
nome = str (input("Digite seu nome: "))
print('-=-'*10)

nota1 = float (input("Digite a nota do primeiro bimestre: "))
nota2 = float (input("Digite a nota do segundo bimestre: "))
nota3 = float (input("Digite a nota do terceiro bimestre: "))
nota4 = float (input("Digite a nota do quarto bimestre: "))
print('-=-'*15)

# Validação das notas

media = (nota1 + nota2 + nota3 + nota4) / 4

if media >= 7:
  print(f"{nome} Sua média é: {media}, você foi aprovado")

elif media <= 5:
  print(f"{nome} Sua média é: {media}, você está reprovado")

else:
  print(f"{nome} Sua média é: {media}, você está de recuperação")

print('-=-'*15)

```
