# 1. Par ou Impar
Peça um número ao usuário e informe se ele é par ou ímpar.

```python
n1 = int(input("Digite um número: "))

if n1 % 2 == 0:
  print (f"O numéro {n1} é par")

else:
  print(f"O número {n1} é impar")
```

# 2. Maior de dois números
Solicite dois números e mostre qual é o maior (ou se são iguais).

```python
n1 = float (input("Digite um número: "))
n2 = float (input("Digite outro número: "))

if n1 > n2:
  print(f"O número {n1} é maior que o número {n2}")

else:
  print(f"O número {n1} é menor que o número {n2}")
```

# 3. Aprovação de aluno
Peça a nota de um aluno e informe:
o Aprovado (nota ≥ 7)
o Recuperação (5 ≤ nota &lt; 7)
o Reprovado (nota &lt; 5)

```python
Nome = (input("Digite seu nome: "))

n1 = float(input("Sua nota do primeiro bimestre é: "))

n2 = float(input("Sua nota do segundo bimestre é: "))

n3 = float(input("Sua nota do terceiro bimestre é: "))

n4 = float(input("Sua nota do quarto bimestre é: "))


media = (n1 + n2 + n3 + n4) / 4

if media >= 7:
  print(f"Sua média no final foi {media} você foi aprovado")

elif media <=5:
  print(f"Sua média no final foi {media} você está reprovado")

else:
  print(f"Sua média no final foi {media} você está de recuperação")
```

# 4. Classificação de idade
Receba a idade e classifique:
o Criança (0–12)
o Adolescente (13–17)
o Adulto (18–59)
o Idoso (60+)

```python
nome = (input("Digite seu nome: "))

idade = int(input("Digite sua idade: "))

if idade <= 12:
  print(f"Você tem {idade} anos de idade, então você é uma criaça")

elif 13 <= idade <= 17:
  print(f"Você tem {idade} anos de idade, então você é um adolescente")

elif 18 <= idade <= 59:
  print(f"Você tem {idade} anos de idade, então você é um adulto")

else:
  print(f"Você tem {idade} anos de idade, então você é um idoso ")
```

# 9. Ano bissexto
Solicite um ano e informe se ele é bissexto ou não.

```python
ano = int(input("Digite o ano: "))

if (ano % 400 == 0) or (ano % 4 == 0 and ano % 100 != 0):
    print(f"{ano} é um ano bissexto.")

else:
    print(f"{ano} não é um ano bissexto.")
```
