# Exercício 1 — Contagem

- Faça um programa que peça um número ao usuário e mostre todos os números de 1 até esse número usando while.

```python
f = int(input("Digite um número: "))

c = 1

while c <= f:
    print(c)
    c += 1
```

# Exercício 4 - Soma dos números
- Peça um número ao usuário e calcule a soma de todos os números de 1 até esse número

```python
numero = int(input("Digite um número inteiro positivo: "))
soma = 0
i = 1

while i <= numero:
    soma += i
    i += 1

print(f"A soma de todos os números de 1 até {numero} é: {soma}")
```

# Exercício 5 — Tabuada

- Peça um número ao usuário e mostre a tabuada desse número de 1 até 10 usando while.

- Exemplo:

- Digite um número: 7


- 7 x 1 = 7

- 7 x 2 = 14

- ...

- 7 x 10 = 70

```python
numero = int(input("Digite um número: "))

contador = 1

while contador <= 10:
    resultado = numero * contador
    print(f"{numero} x {contador} = {resultado}")
    contador += 1
```

#Exercício 6 — Senha

- Crie um programa que peça uma senha ao usuário.

- Enquanto a senha estiver incorreta, mostre:

- Senha incorreta!

- Quando o usuário digitar a senha correta, mostre:

- Acesso permitido!

- Considere que a senha correta seja "1234"

```python
senha_correta = "1234"

senha_digitada = input("Digite a senha: ")

while senha_digitada != senha_correta:
    print("Senha incorreta!")
    senha_digitada = input("Digite a senha novamente: ")

print("Acesso permitido!")
```

#Exercício 7 — Somando até digitar zero

- Faça um programa que peça vários números ao usuário.

- O programa deve continuar pedindo números até que o usuário digite 0.

- Ao final, mostre a soma de todos os números digitados.

- Exemplo:

- Digite um número: 10

- Digite um número: 5

- Digite um número: 8

- Digite um número: 0

```python
soma = 0

while True:
    numero = int(input("Digite um número: "))
    if numero == 0:
        break
    soma += numero

print(f"Soma: {soma}")
```

# Exercício 8 — Maior número

- Peça números ao usuário até que ele digite 0.

- Ao final, mostre qual foi o maior número digitado.

- Exemplo:

- Digite um número: 15

- Digite um número: 8

- Digite um número: 32

- Digite um número: 20

- Digite um número: 0


- Maior número: 32

```python
maior = 0
while True:
  n = int(input('Digite um número: '))
  if n > maior:
    maior = n
  elif n == 0:
    break
print(maior)


```

# Exercício 9 — Média das notas

- Faça um programa que peça 5 notas de um aluno usando while.

- Ao final, calcule e mostre a média das notas.

- Exemplo:

- Digite a nota: 7

- Digite a nota: 8

- Digite a nota: 6

- Digite a nota: 9

- Digite a nota: 10


- Média: 8.0

```python
contador = 5
conta = 0
while contador != 0:
  n = int(input('Digite um número: '))
  conta += n
  contador -= 1
print (conta / 5)
```

#Exercício 10 — Caixa eletrônico

- Faça um programa que comece com um saldo de R$ 1000,00.

- Mostre um menu:

- 1 - Depositar

- 2 - Sacar

- 3 - Ver saldo

- 4 - Sai

```python
saldo = 1000.00

while True:
  print('Caixa Eletrônico')
  print('Opção 1 - Depositar')
  print('Opção 2 - Sacar')
  print('Opção 3 - Ver Saldo')
  print('Opção 4 - Sair')

  n = int(input("Digite uma opção: "))
  if n == 1:
    valor = int(input('Digite o valor desejado para deposito: '))
    saldo += valor
    print(f'Seu saldo agora é de {saldo}')
  elif n == 2:
    valor = int(input('Digite o valor desejado para sacar: '))
    saldo -= valor
    print(f'Seu saldo agora é de {saldo}')
  elif n == 3:
    print(f'Seu saldo é de: {saldo}')
  elif n == 4:
    print('Tchauzinho!!')
    break
  else:
    print('Digite uma opção valida!')
```
