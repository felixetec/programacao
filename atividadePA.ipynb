```python
def somar():
    resultado = 10 + 20
    print(resultado)

somar()
```

```python
def cumprimentar(nome):
    print("Olá,", nome + "!")

cumprimentar("Maria")
cumprimentar("João")
cumprimentar("Lucas")
```

```python
def dobro(numero):
    print("Número:", numero)
    print("Dobro:", numero * 2)

dobro(5)
dobro(10)
dobro(7)
```

```python
def mostrar_idade(idade):
    print("Você tem", idade, "anos.")

mostrar_idade(15)
mostrar_idade(18)
mostrar_idade(20)
```

```python
def verificar_numero(numero):
    if numero > 0:
        print("O número é positivo.")
    elif numero < 0:
        print("O número é negativo.")
    else:
        print("O número é zero.")

verificar_numero(10)
verificar_numero(-5)
verificar_numero(0)
```

```python
def calcular_media(nota1, nota2):
    media = (nota1 + nota2) / 2
    print("Média:", media)

calcular_media(8, 6)
```
